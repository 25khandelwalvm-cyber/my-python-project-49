```python
class Student:
    # Constructor to initialize roll number, name, and marks
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks
        self.total = sum(marks)
        self.percentage = self.total / 5

    # Method to check and display admission status
    def check_admission(self):
        print(f"\n--- Student Result ---")
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Total Marks: {self.total} / 500")
        print(f"Percentage: {self.percentage}%")

        # Check eligibility condition
        if self.percentage > 85:
            print("Status: Congratulations! You can take admission.")
        else:
            print("Status: Sorry, you cannot take admission.")

# --- Main Program ---
# 1. Take student details as input
roll = input("Enter Roll Number: ")
student_name = input("Enter Name: ")

# 2. Take 5 paper marks using a loop
student_marks = []
print("Enter marks for 5 papers (out of 100):")
for i in range(1, 6):
    mark = float(input(f"Paper {i}: "))
    student_marks.append(mark)

# 3. Create the object (this triggers the constructor)
s1 = Student(roll, student_name, student_marks)

# 4. Call the method to display results
s1.check_admission()
```

```python
class Student:
    def __init__(self):
        self.roll_no = ""
        self.name = ""
        self.marks = []
        self.total = 0
        self.percentage = 0.0

    def accept_details(self):

        self.roll_no = input("Enter Roll Number: ")
        self.name = input("Enter Student Name: ")

        print("Enter marks for 5 papers (out of 100):")
        self.marks = []
        for i in range(1, 6):
            while True:
                try:
                    mark = float(input(f"  Paper {i}: "))
                    if 0 <= mark <= 100:
                        self.marks.append(mark)
                        break
                    else:
                        print("  Please enter valid marks between 0 and 100.")
                except ValueError:
                    print("  Invalid input. Please enter a number.")

    def calculate_results(self):
        """Calculates total marks and percentage."""
        self.total = sum(self.marks)
        self.percentage = (self.total / 500) * 100

    def check_admission_status(self):
        """Displays the student's details and admission eligibility."""
        print("\n" + "="*30)
        print("       STUDENT REPORT       ")
        print("="*30)
        print(f"Roll Number:  {self.roll_no}")
        print(f"Name:         {self.name}")
        print(f"Total Marks:  {self.total} / 500")
        print(f"Percentage:   {self.percentage:.2f}%")
        print("-"*30)

        # Admission logic check
        if self.percentage > 85:
            print("Status:       Admission Approved! 🎉")
        else:
            print("Status:       Admission Approved (Via Alternative Route/Regular Quota) 📝")
        print("="*30)

# --- Main Program Execution ---
if __name__ == "__main__":
    # Creating an object of the Student class
    student_obj = Student()

    # Calling class methods using the object
    student_obj.accept_details()
    student_obj.calculate_results()
    student_obj.check_admission_status()
```

```python
# Define the input string
s = "ashish"

# Initialize an empty string to store the reversed result
reversed_s = ""

# Loop through the string backwards from the last index to 0
for i in range(len(s) - 1, -1, -1):
    reversed_s += s[i]

# Print the result
print(f"Original string: {s}")
print(f"Reversed string: {reversed_s}")
```
